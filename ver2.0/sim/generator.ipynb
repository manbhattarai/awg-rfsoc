```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib qt
from scipy.signal import sawtooth

DAC_SR = 10e9 #in GHz
DAC_amplitude = 2**15-1 #16 bit representation in 2's complement format
N_points = 2**20
MAX_POINTS = int(2**16)
t = 1/DAC_SR*np.arange(N_points)

#Generating test signal
freq = 1.8e4 #in Hz
mysignal = 0.9*DAC_amplitude*np.sin(2*np.pi*freq*t)
#mysignal = 0.8*DAC_amplitude*sawtooth(2*np.pi*freq*t)
mysignal = np.int16(mysignal)


plt.subplot(121)
plt.plot(t*1e6,mysignal)
plt.xlabel(r"Time ($u$s)")
plt.ylabel('Amplitude (arb units)')
plt.title("Signal generated to send to the AWG")

with open("data_points.txt", "w") as txt_file:
    for item in mysignal:
        val = f"{item & 0xFFFF:04X}"
        txt_file.write(val+"\n")
```

```python
filename = 'output.txt'
out_vals = []
bit_width=16
with open(filename, 'r') as file:
        for line in file:
            parts = line.split()
            for part in parts:
                try:
                    hex_value = int(part, 16)
                    max_value = 1 << (bit_width - 1)
                    if hex_value >= max_value:
                        hex_value -= (1 << bit_width)
                    out_vals.append(hex_value)
                except ValueError:
                    continue
```

```python
plt.subplot(122)
t=np.arange(len(out_vals))*1/DAC_SR*1e6 #microseconds
plt.plot(t,out_vals,'-o')
plt.xlabel(r"Time ($u$s)")
plt.ylabel('DAC output (unscaled)')
plt.title("Output from AWG generated with simulation")
```
